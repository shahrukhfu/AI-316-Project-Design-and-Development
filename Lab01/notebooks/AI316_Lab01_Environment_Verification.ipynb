### **Project Overview**
The **Industrial Fruit Defect Detection System** is an end-to-end computer vision and deep learning quality control framework designed for high-speed conveyor belts in automated agricultural sorting and food processing facilities. The pipeline combines **classical digital image processing** for feature extraction and surface integrity analysis with **state-of-the-art YOLO object detection** for real-time localization and defect triage.

### **Core Capabilities & Inspection Workflow**
- **High-Throughput Spatial Standardization**: Standardizes variable camera resolutions to fixed dimensions with anti-aliasing downsampling.
- **Photometric & Noise Conditioning**: Multi-stage Gaussian and Median filtering to attenuate specular glare and conveyor reflections.
- **Structural Edge & Contour Profiling**: Dual-threshold Canny edge detection and Otsu bimodal segmentation to isolate skin tears, bruising, and lesions.
- **Real-Time Deep Learning Detection**: YOLO-based bounding box localization, confidence scoring, and multi-class classification.
- **Automated Diagnostic Reporting**: Side-by-side multi-spectral visual grid generation and structured conveyor manifests.
